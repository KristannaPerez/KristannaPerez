Este programa implementa el metodo de eliminacion gaussiana con pivoteo parcial escalado para resolver un sistema de ecuaciones lineales de la forma $A•x=b$

Usamos la formula del multiplicador para eliminar un elemento $a_{i,j}$ \\
$m=\frac{a_{j,i}}{a_{i,i}}$

Despues hacemos la seleccion del pivote usando la formula de \\
$max=(\frac{a_{r,i}}{s_r})$

Cuando la matriz aumentada se ah transformado en forma triangular superior la solucion se obtiene resolviendo las ecuaciones una por una, empezando por la ultima asi aplicamos la sustitucion hacia atras

Antes de comenzar la eliminacion se verifica que ninguna escala sea cero ya que esto indicaria que no hay solucion unica

```python
import numpy as np

def eliminacion_gaussiana_pivoteo_escalado(A, b):
    n = len(b)  # Número de ecuaciones
    # Crear la matriz aumentada
    matriz_aumentada = np.hstack([A, b.reshape(-1, 1)])

    # Inicializar el vector de índices de las filas y las escalas
    NROW = list(range(n))  # Índices de las filas
    escalas = [max(abs(fila)) for fila in A]  # Escala máxima de cada fila

    # Verificar si alguna escala es cero
    if any(escala == 0 for escala in escalas):
        raise ValueError("No existe solución única (una escala es cero).")

    # Eliminación hacia adelante
    for i in range(n - 1):
        # Selección del pivote según el pivoteo parcial escalado
        fila_pivote = max(
            range(i, n),
            key=lambda r: abs(matriz_aumentada[NROW[r], i]) / escalas[NROW[r]]
        )
        if abs(matriz_aumentada[NROW[fila_pivote], i]) == 0:
            raise ValueError("No existe solución única (la matriz es singular).")

        # Intercambio de filas
        if NROW[i] != NROW[fila_pivote]:
            NROW[i], NROW[fila_pivote] = NROW[fila_pivote], NROW[i]

        # Eliminación de los elementos debajo del pivote
        for j in range(i + 1, n):
            m = matriz_aumentada[NROW[j], i] / matriz_aumentada[NROW[i], i]
            m = round(m, 3)  # Redondeo de multiplicadores a 3 dígitos
            matriz_aumentada[NROW[j], i:] -= m * matriz_aumentada[NROW[i], i:]
            matriz_aumentada[NROW[j], i:] = np.round(matriz_aumentada[NROW[j], i:], 3)

    # Sustitución hacia atrás
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        suma = sum(matriz_aumentada[NROW[i], j] * x[j] for j in range(i + 1, n))
        x[i] = (matriz_aumentada[NROW[i], -1] - suma) / matriz_aumentada[NROW[i], i]
        x[i] = round(x[i], 3)  # Redondeo del resultado a 3 dígitos

    return x


# Ejemplo 3
A = np.array([[2.11, -4.21, 0.921],
              [4.01, 10.2, -1.12],
              [1.09, 0.987, 0.832]])
b = np.array([2.01, -3.09, 4.21])

# Resolver el sistema
solucion = eliminacion_gaussiana_pivoteo_escalado(A, b)

print("Solución del sistema:", solucion)
```
