Eliminacion gaussiana con sustitucion hacia atras

Si el pivote actual es cero buscamos una fila con diferente de cero, esto permite continuar con la eliminacion.

Hacemos la reduccion de filas con las formulas: \\
$m_{ji}=\frac{a_{ji}}{a_{ii}}$\\
$F_j=F_j-m_{ji}•F_i$

Una vez que la matriz es triangular superior, resolvemos el sistema de abajo hacia arriba con la formula:\\
$x_i=\frac{b_{ij}-Σ_{j=i+1}a_{ij}x_j}{a_{ii}}$

Si un pivote es cero el sistema no tiene solucion unica.

```python
import numpy as np

def eliminacion_gaussiana(A, b):
    n = len(A)
    # Matriz aumentada
    Ab = np.hstack((A, b.reshape(-1, 1)))

    # Proceso de eliminación
    for i in range(n):
        # Paso 2: Verificar si el pivote es cero, buscar intercambio
        if Ab[i, i] == 0:
            for p in range(i + 1, n):
                if Ab[p, i] != 0:
                    Ab[[i, p]] = Ab[[p, i]]  # Intercambio de filas
                    break
            else:
                return "No existe una solución única"

        # Paso 4: Escalonar la matriz
        for j in range(i + 1, n):
            m = Ab[j, i] / Ab[i, i]
            Ab[j, i:] -= m * Ab[i, i:]

    # Verificar si el sistema es compatible y determinado
    if Ab[n - 1, n - 1] == 0:
        return "No existe una solución única"

    # Sustitución hacia atrás
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (Ab[i, -1] - np.dot(Ab[i, i + 1:n], x[i + 1:n])) / Ab[i, i]

    return x


# Ejemplo 2
A = np.array([[1, -1, 2, -1],
              [2, -2, 3, -3],
              [1, 1, 1, 0],
              [1, -1, 4, 3]], dtype=float)

b = np.array([-8, -20, -2, 4], dtype=float)

resultado = eliminacion_gaussiana(A, b)

if isinstance(resultado, str):
    print(resultado)
else:
    print("Solución:", resultado)
```
