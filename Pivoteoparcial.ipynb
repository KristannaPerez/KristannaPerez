Eliminacion Gaussiana con pivoteo parcial

Primero se forma la matriz aumentada que combina la matriz de coeficientes A con el vector de terminos independiente B. \\
Despues el pivoteo parcial busca el elemento de mayor valor absoluto en la columna actual, luego se intercambian las filas si el mayor elemento no esta en la fila actual.

El objetivo es transformar la matriz A en una matriz triangular superior donde todos los elementos debajo de la diagonal principal son cero.

Cuando la matriz A se ah triangularizado el sistema se resuelve desde la ultima ecuacion hacia la primera y se calculan con: \\
$x_i=\frac{1}{a_{ii}}(b_i-Σ_{j=i+1}a_{ij}x_j)$

```python
import numpy as np

def eliminacion_gaussiana_con_pivoteo_parcial(A, b):
    n = len(A)
    # Crear la matriz aumentada
    Aug = np.hstack((A, b.reshape(-1, 1)))

    # Eliminación gaussiana con pivoteo parcial
    for i in range(n):
        #Encontrar la fila con el mayor elemento en la columna i
        fila_max = i + np.argmax(abs(Aug[i:, i]))

        #Verificar si el sistema no tiene solución única
        if Aug[fila_max, i] == 0:
            return "No existe solución única"

        #Intercambiar filas si es necesario
        if fila_max != i:
            Aug[[i, fila_max]] = Aug[[fila_max, i]]

        #Proceso de eliminación
        for j in range(i + 1, n):
            multiplicador = Aug[j, i] / Aug[i, i]
            Aug[j, i:] -= multiplicador * Aug[i, i:]

    #Sustitución hacia atrás
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (Aug[i, -1] - np.sum(Aug[i, i+1:n] * x[i+1:n])) / Aug[i, i]

    return x

# Ejemplo del libro
A = np.array([[0.003, 59.14],
              [5.291, -6.13]])
b = np.array([59.17, 46.78])

solucion = eliminacion_gaussiana_con_pivoteo_parcial(A, b)
print("Solución del sistema:", solucion)
```
